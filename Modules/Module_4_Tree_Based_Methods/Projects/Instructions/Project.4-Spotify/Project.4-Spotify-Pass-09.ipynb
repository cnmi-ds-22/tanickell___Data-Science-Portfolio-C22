## Problem Definition


#### Business Problem

This project aims to predict the popularity of unreleased songs shortly following release, based on song data from Spotify's Top 200 Weekly (Global) charts from 2020 and 2021. Companies like Spotify may wish to understand and predict what types of tracks become hits and go viral, as these tracks can be promoted better and charge more for advertising.

#### Data Science Problem

Song popularity will be predicted using one of three different tree-based regression models: decision-tree, random forest, and xgboost.

Popularity is a continuous, numeric floating-point value. As such, this is a supervised regression problem, where the target of the prediction is the upcoming popularity of a prospective track release.

### XGBoost Regression

```python
features = df.drop(target_col, axis = 1)
target = df[target_col]
X = features
y = target
```

```python
# Multiple CV — loop count
n = 30

# Create array to hold results
rmses = np.zeros(n)

# Multiple CV loop
for i in range(n):

  # Train Test Split
  X_train, X_test, y_train, y_test = train_test_split( X, y, test_size = 0.15 )

  # Create an empty model
  model = xgb.XGBRegressor(objective ='reg:squarederror', verbosity=0, seed = 10)

  # Train ( fit ) the model
  model.fit(X_train, y_train)

  # Make predictions using the model
  y_pred = model.predict(X_test)

  # Calculate the metric and save to array
  rmses[i]  = np.sqrt(np.mean(np.square(y_test - y_pred)))
```

```python
# Calculate mean of results
mean_rmse = rmses.mean()
mean_rmse
```

### How do we choose the parameters?

```python
# Specify the parameters you want to try and their ranges.
param_test = {
 'max_depth'     : [ 3, 4, 5, 6, 7 ],
 'learning_rate' : [ 0.1, 0.2, 0.3, 0.4 ],
 'n_estimators'  : [ 20, 40, 60, 80, 100, 120, 140 ],
}

# Perform the grid search
gsearch = GridSearchCV(
    estimator = xgb.XGBRegressor( objective = 'reg:squarederror', seed = 10 ),
    param_grid = param_test,
    scoring = 'neg_mean_squared_error',
    cv = 5,
)

# Fit to training data
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.05, random_state = 42 )
model = gsearch.fit(X_train,y_train)

# See grid search results
print(model.best_params_)


```

```python
params = model.best_params_

# Multiple CV — loop count
n = 30

# Create array to hold results
rmses = np.zeros(n)

# Multiple CV loop
for i in range(n):

  # Train Test Split
  X_train, X_test, y_train, y_test = train_test_split( X, y, test_size = 0.15 )

  # Create an empty model
  model = xgb.XGBRegressor(
      objective = 'reg:squarederror',
      verbosity=0,
      **params,
      seed = 10)

  # Train ( fit ) the model
  model.fit(X_train, y_train)

  # Make predictions using the model
  y_pred = model.predict(X_test)

  # Calculate the metric and save to array
  rmses[i]  = np.sqrt(np.mean(np.square(y_test - y_pred)))
```

```python
# Calculate mean of results
mean_rmse = rmses.mean()
mean_rmse
```

  - n_estimators: 40
- XGBoost yields a good result: RMSE comparable to Decision Tree and Random Forest (10.6 and 10.4, respectively)
- Average song popularity = 70. For songs at this value, the model should predict somewhere between 59.5 and 80.5 (70 ± 10.51).
- Few strong signals dominate; others noise
- change in rmse with tree depth indicates meaningful signal -- real predictive patterns in data
- As with both Decision Tree and Random Forest, relative to linear regression, XGBoostt has the advantage of not needing to meet the requirements of linear regression (features are independent from the target), and are independent from one another (not collinear).
- XGBoost has slightly higher error (RSME) than Random Forest, but is much faster to compute RSMEs. By contrast, XGBoost has a slightly lower RSME than Decision Tree, but requires a little more setup to implement.
