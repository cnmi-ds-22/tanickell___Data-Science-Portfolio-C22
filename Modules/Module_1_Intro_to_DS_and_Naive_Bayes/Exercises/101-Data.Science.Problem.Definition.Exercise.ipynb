# Data Science Problem Definition


For each item develop a well-worded data science problem statement.  Your answer should include what type of data science problem this is ( supervised vs unsupervised ) and then whether it is a classification, regression, or clustering problem.  If it is a supervised problem, you should state what the target of the prediction is.

1. Zillow wants you to predict real estate sales prices. They have a large
database of 12 years of home features accompanied by the corresponding
sales prices.


Statement:

Zillow wants to optimize sales by predicting how much a new listed house can sell for based on its features (age, renovations, flooring, etc.).

This is a *supervised regression* data science problem (numerical, continuous), where the target of the prediction is home values (in $).



---

2. Amazon wants you to develop a code that upsells (add additional items to the
customer’s order) based on what items have been viewed and bought by that
customer over the past 2 years.



Statement:

Amazon wants to drive sales by predicting what else a customer might purchase within a given order and market it to them on-page as they go through the purchase process.

This is an *unsupervised clustering* data science problem, as the targets of the prediction, the potential merchandise, are not known.



---

3. MGM wants to know what features of a movie seem to increase the movie’s
box office income. They have a large database of previously released movies
accompanied by the resulting box office revenue.



Statement:

MGM wants to improve ticket sales (box office income) and drive revenue by predicting what types of movies will perform better based on their features (leading actor, genre, length, etc.).

This is a *supervised regression* (numerical continuous) data science problem, where the target of the prediction is sales revenue (in $).



---

4. Spotify wants you to develop a method of predicting whether a customer is
about to terminate their service BEFORE they terminate so that they can offer
enticements to the customer to keep the service and not terminate (called
‘churn’)



Statement:

Spotify wishes to predict whether a customer is currently 'high risk' of terminating their service based on existing customer data (how active they are, what types of music they listen to, engagement with product features, etc.).

This is a *supervised binary classification* data science problem (categorical nominal), where the target of the prediction is whether a customer is at risk of terminating their subscription.


---

5. Comcast wants you to develop a chat bot (automated customer service) to
take text input and respond with the appropriate help and/or service
response.



Statement:

Comcast is hoping to increase their customer service offerings by providing unstructured assistance to customers based on their text input.

This is an *unsupervised clustering* data science problem, as the target of the prediction is text output that is grouped with text input (i.e., customer queries), and is otherwise undefined and unlabeled.



---

6. Instagram wants you to analyze images and output whether the image
contains people, animals, scenery, or some other subject.


Statement:

Instagram wants to develop an image recognition service to promote it as a new feature and drive new signups.

This is a *supervised classification* data science problem (categorical nominal), where the target of the prediction is the type of subject in a given image.
