# Data Science Problem Definition


For each item develop a well-worded data science problem statement.  Your answer should include what type of data science problem this is ( supervised vs unsupervised ) and then whether it is a classification, regression, or clustering problem.  If it is a supervised problem, you should state what the target of the prediction is.

1. Zillow wants you to predict real estate sales prices. They have a large
database of 12 years of home features accompanied by the corresponding
sales prices.


Statement:

Zillow wants to optimize future home sales by using past home sale prices and housing features, such as furnishings, renovations, square footage, etc., to develop a sales model. This is a *supervised regression* data science problem, whose prediction target is home values (in $).



---

2. Amazon wants you to develop a code that upsells (add additional items to the
customer’s order) based on what items have been viewed and bought by that
customer over the past 2 years.



Statement:

Amazon wants to determine items to suggest for user orders based on their viewing and purchase history. This is an *unsupervised clustering* data science problem that relates unknown data — item suggestions — to existing user data.



---

3. MGM wants to know what features of a movie seem to increase the movie’s
box office income. They have a large database of previously released movies
accompanied by the resulting box office revenue.



Statement:

MGM wants to model box office revenue based on data (genre, length, directors/actors, etc.) related to previously released movies. This is a *supervised regression* data sciente problem, with a prediction target of potential box office sales.



---

4. Spotify wants you to develop a method of predicting whether a customer is
about to terminate their service BEFORE they terminate so that they can offer
enticements to the customer to keep the service and not terminate (called
‘churn’)



Statement:

Spotify is looking to predict whether users are stable or unstable (that is, in a decision state to unsubscribe to services). This is a *supervised binary classification* data science problem, where the target is the number of churns (i.e., success is that this value decreases relative to a reference point).

(Note here that this could also be considered an *unsupervised clustering* problem when looking only at factors relating user behavior to other behaviors.)


---

5. Comcast wants you to develop a chat bot (automated customer service) to
take text input and respond with the appropriate help and/or service
response.



Statement:

Comcast wants to provide improved chatbot services to increase their customer support options, where the chatbot provides text-based suggestions based on text input from the user. This is an *unsupervised clustering* data science problem.



---

6. Instagram wants you to analyze images and output whether the image
contains people, animals, scenery, or some other subject.


Statement:

Instagram wants to implement pattern-recognition features in its image analysis tools that can discern contents of different types from uploaded pictures. This is a *supervised classification* problem, where the prediction target is a label for the contents of the image (animal, garden, car, etc.).
