## Working with lists

1. Use a variable to represent a famous person’s name and then print that person’s name in lowercase, uppercase and title case.

```python
famous_person = "Francis Bacon"
print(famous_person.lower())
print(famous_person.upper())
print(famous_person.title())
```

2. Print a message about the person you defined above using an f-string.

```python
famous_birthday = "January 22"

print(f"{famous_person} was born on {famous_birthday}.")
```
