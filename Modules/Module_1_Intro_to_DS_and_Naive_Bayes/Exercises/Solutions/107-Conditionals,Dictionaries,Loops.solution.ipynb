4. Write a python program that will take a word and print a message stating whether that word starts with a vowel or a consonant. Test it out with several words.

Example input: ‘popcorn’

Example output: ‘This word starts with a consonant.’


```python
# Solution
words = [ 'popcorn', 'apple' ]

vowels = 'aeiouAEIOU'
for word in words:
  if word[0] in vowels:
    print(f"{word}: This word starts with a vowel")
  else:
    print(f"{word}: This word starts with a consonant.")
```

```python
# Solution - Variant
words = [ 'popcorn', 'apple' ]
vowels = { x: "vowel" for x in 'aeiouAEIOU' }

for word in words:
  print(f"{word}: This word starts with a { vowels.get(word[0],'consonant') }")
```

5. Write a python program that will loop through the following list:
`pets = ['Muffin', 'Kiki', 'Moe', 'Daisy']`. If the length of the pet's name is less than 5 characters, print out the name.  

```python
# Solution
pets = ['Muffin', 'Kiki', 'Moe', 'Daisy']
for pet in pets:
  if len(pet) < 5:
    print(pet)
```

```python
# Solution - Variant
pets = ['Muffin', 'Kiki', 'Moe', 'Daisy']
print("\n".join( pet for pet in pets if len(pet) < 5 ) )
```

1. Create the following two lists of numbers:

`list_1 = [1,2,3,4,5,6,7,8]`

`list_2 = [9,10,11,12,13,14]`

 Write a program to create a new list such that the new list should contain odd numbers from the first list and even numbers from the second list.

```python
# Solution
list_1 = [1,2,3,4,5,6,7,8]
list_2 = [9,10,11,12,13,14]

list_3 = []
for num in list_1:
  if ( num % 2 != 0 ): # if not divisible by 2, i.e. odd
    list_3.append(num)

for num in list_2:
  if ( num % 2 == 0 ): # if divisible by 2, i.e. even
    list_3.append(num)

print(list_3)
```

```python
# Solution - Variant
list_1 = [1,2,3,4,5,6,7,8]
list_2 = [9,10,11,12,13,14]

list_3 = (
    [ num for num in list_1 if ( num % 2 != 0 ) ] +
    [ num for num in list_2 if ( num % 2 == 0 ) ]
    )
print(list_3)
```

2. Write a program that will check to see if a username contains at least one of the following:

*   A number
*   A letter
*   A symbol ($, %, ! or ?)

It should print out 'Good password' if it meets the above requirements and 'Bad password' otherwise.

Example input: 'Hello5?'

Example output: 'Good password'

Hint: `import string` and then use `list(string.ascii_lowercase)` to get a list of all lowercase letters


```python
# Solution - Variant
import string

num = list(string.digits)
letters = list(string.ascii_letters)
symbols = list(string.punctuation)

password = 'Hello5?'

num_check = let_check = sym_check = False

for letter in password.lower():
  if letter in num:
    num_check = True
  elif letter in letters:
    let_check = True
  elif letter in symbols:
    sym_check = True

if num_check and let_check and sym_check:
  print('Good Password')
else:
  print('Bad Password')
```

```python
# Solution - Variant
import string

password = 'Hello5?'

lookup = (
  { x:"d" for x in string.digits } |
  { x:"a" for x in string.ascii_letters } |
  { x:"p" for x in string.punctuation }
)

tf = ( len(set( lookup.get(c,"_") for c in password ) - set("_") ) == 3 )
print(('Bad Password', 'Good Password')[tf])
```

```python
# Solution - Variant
import string

password = 'Hello5?'

d, s, p = set(string.digits), set(string.ascii_letters), set(string.punctuation)

x = set(password)
print(('Bad Password', 'Good Password')[all([ (x & i) for i in (d,s,p)])])
```

```python
import string

password = 'Hello5?'

reqs = [
  string.punctuation,
  string.ascii_letters,
  string.digits,
]

tf = all( set(password).intersection(req) for req in reqs )
print(('Bad Password', 'Good Password')[tf])

```
