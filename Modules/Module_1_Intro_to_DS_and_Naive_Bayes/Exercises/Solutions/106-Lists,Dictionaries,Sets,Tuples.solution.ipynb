Starting with a given list, count the occurrences of each unique element.  Create a dictionary to store the count of each element.

````python
list1 = [34, 544, 544, 34, 67, 67, 34, 5, 7, 9]
````

````python
# Solution
value_counts = {}
for i in list1:
  value_counts[i] = value_counts.get(i,0) + 1
value_counts
````

````python
# Solution - variant
value_counts = {}
for i in set(list1):
  value_counts[i] = list1.count(i)
value_counts
````

````python
# Solution - variant
value_counts = { i: list1.count(i) for i in set(list1) }
value_counts
````

Given two sets, find the intersection of the sets and remove those elements from the first set

```python
set1 = {x for x in range(20)}
set2 = {x for x in range(3,10)}
```


````python
# Solution
set1 = {x for x in range(20)}
set2 = {x for x in range(3,10)}

set1 -= set1.intersection(set2)
set1
````

````python
# Solution - variant
set1 = {x for x in range(20)}
set2 = {x for x in range(3,10)}

set1 = set1 - set2
set1
````

````python
# Solution - variant
set1 = {x for x in range(20)}
set2 = {x for x in range(3,10)}

set1 -= set2
set1
````

````python
# Solution - variant
set1 = {x for x in range(20)}
set2 = {x for x in range(3,10)}

set1.difference_update(set2)
set1
````

Given a list check if each element already exists in a dictionary as a key's value.  If not, delete the element from the list.

```python
list0 = [4, 5, 6, 7, 3, 2, 5, 4, 1, 3]
fruit_counts = {
    "apple" : 4,
    "orange": 8,
    "grapefruit": 3,
    "grapes": 7,
    }

```

````python
fruit_counts = {
    "apple" : 4,
    "orange": 8,
    "grapefruit": 3,
    "grapes": 7,
    }
fruit_counts
````

````python
# Solution
list0 = [4, 5, 6, 7, 3, 2, 5, 4, 1, 3]

# begin with the end in mind: pop from left to right, but count from right to left
values = fruit_counts.values()
for i in range(-len(list0), 0):
  if list0[i] in values:
    list0.pop(i)
list0
````

````python
# Solution - variant
list0 = [4, 5, 6, 7, 3, 2, 5, 4, 1, 3]

# pop from right to left, but count from left to right
values = fruit_counts.values()
x = len(list0)-1
for i, v in enumerate(list0[::-1]):
  if v in values:
    list0.pop(x-i)
list0
````

````python
# Solution - variant
list0 = [4, 5, 6, 7, 3, 2, 5, 4, 1, 3]

list0 = [ v for v in list0 if v not in set(fruit_counts.values()) ]
list0
````

Given a string input calculate and display the number of digits and letters

```python
s = "I <3 python"
```

````python
alphabet = set("abcdefghijklmnopqrstuvwxyz")
len(alphabet)
````

````python
numbers = set("0123456789")
len(numbers)
````

````python
# Solution
s = "I <3 python"

count = { "letters": 0, "digits": 0 }
for c in s.lower():
  if c in alphabet: count["letters"] += 1
  if c in numbers: count["digits"] += 1

count
````

````python
# Solution - variant
s = "I <3 python"

count = { "letters": 0, "digits": 0, "_": 0 }
ld = {
  **{ x:"letters" for x in alphabet },
  **{ x:"digits" for x in numbers }
}
for c in s.lower(): count[ld.get(c,"_")] += 1
count.pop("_")
count
````

````python
# Solution - variant
from collections import Counter

s = "I <3 python"

dict(Counter(
  'letters' if c.isalpha() else
  'digits' if c.isdigit() else
  'other'
    for c in s
  )
)
````

````python
# Solution - variant
s = "I <3 python"

{
  'letters': sum( c.isalpha() for c in s ),
  'digits' : sum( c.isdigit() for c in s ),
}
````
