# Problem 3:

An artist competition has entries from three painters: Pam, Paul and Peter.

- Pam put in 15 paintings, 4% of her works have won First Prize.
- Paul put in 5 paintings, 5% of his works have won First Prize.
- Peter put in 10 paintings, 3% of his works have won First Prize.

What is the probability that Pam will win First Prize?

## Using Python with Synthetic Data


```python
# Import needed modules
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from fractions import Fraction
```

### Choose size of population

```python
n = 30_000
f"{n:,}"
```

### Empty data frame

```python
df = pd.DataFrame()
df
```

### Create the population, i.e. the number of paintings

```python
df["id"] = np.arange(n)
df[:5]
```

```python
df.shape
```

### Tag with the owner of the painting


```python
df["owner"] = np.repeat("Pam Paul Peter".split(), [15_000, 5_000, 10_000])
df
```

```python
df[["owner"]].value_counts(normalize=True).sort_index()*100
```

```python
df[["owner"]].value_counts().sort_index()
```

### Tag those that won first prize

```python
# set everyone to "show"
df["prize"] = "show"
df[:5]
```

```python
df[["owner","prize"]].value_counts().sort_index()
```

```python
# adjust 4% of Pam's to "first"
who = ( df["owner"] == "Pam" )
who_first = df[ who ].sample(frac=0.04).index
df.loc[ who_first, "prize" ] = "first"
# df.iloc[ who_first, df.columns.get_loc("prize") ] = "first"
df[["owner","prize"]].value_counts().sort_index()
```

```python
# adjust 5% of Paul's to "first"
who = ( df["owner"] == "Paul" )
who_first = df[ who ].sample(frac=0.05).index
df.loc[ who_first, "prize" ] = "first"
# df.iloc[ who_first, df.columns.get_loc("prize") ] = "first"
df[["owner","prize"]].value_counts().sort_index()
```

```python
# adjust 3% of Peter's to "first"
who = ( df["owner"] == "Peter" )
who_first = df[ who ].sample(frac=0.03).index
df.loc[ who_first, "prize" ] = "first"
# df.iloc[ who_first, df.columns.get_loc("prize") ] = "first"
df[["owner","prize"]].value_counts().sort_index()
```

```python
vc_ds = df[["owner","prize"]].value_counts().sort_index()
vc_ds
```

```python
firsts = vc_ds[:,"first"]
firsts
```

```python
firsts/firsts.sum() * 100
```
