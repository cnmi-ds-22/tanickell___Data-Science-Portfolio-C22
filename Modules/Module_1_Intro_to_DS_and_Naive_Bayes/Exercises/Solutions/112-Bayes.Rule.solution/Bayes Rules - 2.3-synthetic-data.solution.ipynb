# Problem 2


Beth is planning a picnic today, but the morning is cloudy.

- 50% of all rainy days start off cloudy
- Cloudy mornings are common (40% of all days start cloudy)
- This is usually a dry month (10% of days tend to be rainy)

What is the chance of rain during the day when it starts of cloudy?

## Using Python with Synthetic Data


```python
# Import needed modules
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from fractions import Fraction
```

### Choose size of population

```python
n = 1_000
f"{n:,}"
```

### Empty data frame

```python
df = pd.DataFrame()
df
```

### Create the population, i.e. the number of days

```python
df["id"] = np.arange(n)
df[:5]
```

```python
df.shape
```

### Tag with the rainy days


```python
df["rainy"] = np.repeat("yes no".split(), [100,900])
df
```

```python
df[["rainy"]].value_counts(normalize=True).sort_index()*100
```

```python
df[["rainy"]].value_counts().sort_index()
```

### Tag cloudy days

```python
# set cloudy to "no"
df["cloudy"] = "no"
df[:5]
```

```python
df[["rainy","cloudy"]].value_counts().sort_index()
```

```python
# adjust 50% of rainy days to cloudy "yes"
days = ( df["rainy"] == "yes" )
days_rainy = df[ days ].sample(frac=0.50).index
df.loc[ days_rainy, "cloudy" ] = "yes"
# df.iloc[ days_rainy, df.columns.get_loc("cloudy") ] = "yes"
df[["rainy","cloudy"]].value_counts().sort_index()
```

```python
# adjust remainder of not rainy days to cloudy "yes"
days = ( df["rainy"] == "no" )
days_rainy = df[ days ].sample(n=(400-50)).index
df.loc[ days_rainy, "cloudy" ] = "yes"
# df.iloc[ days_rainy, df.columns.get_loc("cloudy") ] = "yes"
df[["rainy","cloudy"]].value_counts().sort_index()
```

```python
vc_ds = df[["rainy","cloudy"]].value_counts().sort_index()
vc_ds
```

```python
rainy = vc_ds[:,"yes"]
rainy
```

```python
rainy/rainy.sum() * 100
```
