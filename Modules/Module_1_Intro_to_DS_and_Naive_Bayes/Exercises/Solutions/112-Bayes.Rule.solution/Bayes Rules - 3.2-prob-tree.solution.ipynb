# Problem 3:

An artist competition has entries from three painters: Pam, Paul and Peter.

- Pam put in 15 paintings, 4% of her works have won First Prize.
- Paul put in 5 paintings, 5% of his works have won First Prize.
- Peter put in 10 paintings, 3% of his works have won First Prize.

What is the probability that Pam will win First Prize?

## Probability tree with arbitrary start number


Assume a collection of 30,000 paintings.
- Of the 30,000, 15,000 are Pam's painting ( = 30,000 * 15/30 )
  - of the 15,000 paintings, 600 have one first prize ( = 15,000 * 4% )
- Of the 30,000, 5,000 are Paul's paintings ( = 30,000 * 5/30 )
  - of the 5,000 paintings, 250 have one first prize ( = 5,000 * 5% )
- Of the 30,000, 10,000 are Peter's painting ( = 30,000 * 10/30 )
  - of the 10,000 paintings, 300 have one first prize ( = 10,000 * 3% )

- How many paintings have one first prize? 1,150 ( =600+250+300 )
- How many first prize paintings are Pam's? 600

- $ 600 \div 1150  = 0.5217 \approx 52\% $
