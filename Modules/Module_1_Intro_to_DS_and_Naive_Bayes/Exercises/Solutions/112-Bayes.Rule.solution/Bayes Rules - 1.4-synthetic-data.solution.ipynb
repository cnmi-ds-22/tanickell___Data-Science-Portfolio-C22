# Problem 1


Joan says she is itchy. There is a test for Allergy to Cats, but this test is not always right:

* For people that really do have the allergy, the test says "Yes" 80% of the time
* For people that do not have the allergy, the test says "Yes" 10% of the time

If 1% of the population has the allergy, and Joan's test says "Yes", what are the chances that Joan really has the allergy?

## Using Python with Synthetic Data


```python
# Import needed modules
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from fractions import Fraction
```

### Choose size of population

```python
n = 10_000
f"{n:,}"
```

### Empty data frame

```python
df = pd.DataFrame()
df
```

### Create the population

```python
df.index = np.arange(n)
df[:5]
```

### Tag those who do have the allergy

```python
# initially, tag everyone as not having the allergy
df["has_allergy"] = "F"

# tag 1% as having the allergy
df.iloc[:n//100, df.columns.get_loc("has_allergy")] = "T"

# display the transition
df[n//100-5:n//100+5]
```

```python
# display frequency
df[["has_allergy"]].value_counts(normalize=True)*100
```

```python
# display absolute counts
df[["has_allergy"]].value_counts()
```

### Tag those that have a positive test

```python
# set everyone to negative
df["test"] = "neg"


# adjust 80% of those with allergies to "pos"
has_allergy = ( df["has_allergy"] == "T" )
ha_pos_idx = df[ has_allergy ].sample(frac=0.8).index
df.loc[ ha_pos_idx, "test" ] = "pos"


# adjust 10% of those without allergies to "pos"
has_allergy = ( df["has_allergy"] == "F" )
ha_pos_idx = df[ has_allergy ].sample(frac=0.1).index
df.loc[ ha_pos_idx, "test" ] = "pos"
```

```python
vc_ds = df.value_counts()
vc_ds
```

#### Bayes


Calculate conditional and marginal probabilities.

```python
p_has_allergy = (vc_ds["T",:].sum())/n
p_has_allergy
```

```python
p_test_positive = (vc_ds[:,"pos"].sum())/n
p_test_positive
```

```python
p_test_pos_given_has_allergy = vc_ds["T","pos"]/(vc_ds["T",:]).sum()
p_test_pos_given_has_allergy
```

Assemble using Bayes theorem.

```python
p_has_allergy_given_test_post = p_test_pos_given_has_allergy * p_has_allergy / p_test_positive
p_has_allergy_given_test_post
```
