# Problem 2


Beth is planning a picnic today, but the morning is cloudy.

- 50% of all rainy days start off cloudy
- Cloudy mornings are common (40% of all days start cloudy)
- This is usually a dry month (10% of days tend to be rainy)

What is the chance of rain during the day when it starts of cloudy?

## Probability tree with arbitrary start number


Assume 1,000 days.
- of those 1,000 days, 100 days are rainy ( = 1,000 * 10% )
  - Of 100 rainy days, 50 start cloudy ( = 100 * 50% )
  - Of 100 rainy days, 50 start not cloudy ( = 100 - 50 )
- of those 1,000 days, 900 days are dry ( = 1,000 * 90% )

- of those 1,000 days, 400 days start cloudy ( = 1,000 * 40% )

- Number of days start cloudy? 400
- Number of rainy days that started cloudy = 50

- $ 50/400 = 0.125 $

- $ = 12.5\% $
