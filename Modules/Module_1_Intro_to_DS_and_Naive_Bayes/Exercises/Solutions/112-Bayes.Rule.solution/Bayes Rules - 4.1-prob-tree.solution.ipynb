
# Problem 4

A man speaks the truth 2 out of 3 times. He throws one die and says that number obtained is a three. Find the probability that the number obtained is actually a three.

The challenge in this problem is that there are three classes hidden in the description:

1. the roll of the die
1. his decision to lie or tell the truth
1. what he says ( most importantly, when he lies )

It's that third class that is the least obvious and, in fact, left to the use to determine its events.

## Probability tree with numbers


### Equal probability

When he lies and he rolls a number that is not a three, he is equally likely to say any number other than what is rolled, including a three.

Pick 90 for the number of rolls. Of those 90, assume a perfect even distribution, i.e. the same count for all numbers.

- How many rolls does he tell the truth? 60 rolls ( $ = \frac{2}{3} \times 90 $ )

  - Of those 60, how many rolls are 3? 10 rolls ( $ = \frac{1}{6} \times 60 $ )

    - Of those 10, how many rolls does he call as "3"? These are the number of rolls where the die is 3 and he says "3": **10 rolls** ( $ = \frac{10}{10} \times 10 $ )

- How many rolls does he lie? 30 rolls ( $ = \frac{1}{3} \times 90 $ )

  - Of those 30 rolls, how many are 3? These are the number of rolls where the die is 3, but he will NOT say "3": 5 rolls ( $ = \frac{1}{6} \times 30 $ )

  - Of those 30 rolls, how many are NOT 3? These are the number of rolls where he may or may not say "3": 25 rolls ( $ = \frac{5}{6} \times 30 $ )

    - ***Assuming an equal probability***, of those 25 rolls, how many does he call as 3? These are the number of rolls where the die is something other than a three, but he says "3": **5 rolls** ( $ = \frac{1}{5} \times 25 $ )

- Of the 90 rolls, on how many rolls does he say "3"? 15 = 10 truth + 5 lies rolls

- Of those 15 rolls, how many are actually a 3? 10

- 10/15 = 2/3


### Lies using 3

When he lies and he rolls a number that is not a three, he always says 3.

Pick 90 for the number of rolls. Of those 90, assume a perfect even distribution, i.e. the same count for all numbers.

- How many rolls does he tell the truth? 60 rolls ( $ = \frac{2}{3} \times 90 $ )

  - Of those 60, how many rolls are 3? 10 rolls ( $ = \frac{1}{6} \times 60 $ )

    - Of those 10, how many rolls does he call as "3"? These are the number of rolls where the die is 3 and he says "3": **10 rolls** ( $ = \frac{10}{10} \times 10 $ )

- How many rolls does he lie? 30 rolls ( $ = \frac{1}{3} \times 90 $ )

  - Of those 30 rolls, how many are 3? These are the number of rolls where the die is 3, but he will NOT say "3": 5 rolls ( $ = \frac{1}{6} \times 30 $ )

  - Of those 30 rolls, how many are NOT 3? These are the number of rolls where he may or may not say "3": 25 rolls ( $ = \frac{5}{6} \times 30 $ )

    - ***Assuming always 3***, of those 25 rolls, how many does he call as 3? These are the number of rolls where the die is something other than a three, but he says "3": **25 rolls** ( $ = \frac{5}{5} \times 25 $ )

- Of the 90 rolls, on how many rolls does he say "3"? 35 = 10 truth + 25 lies rolls

- Of those 35 rolls, how many are actually a 3? 10

- 10/35 = 2/7


### Lies using something other than 3

When he lies and he rolls a number that is not a three, he never says 3.

Pick 90 for the number of rolls. Of those 90, assume a perfect even distribution, i.e. the same count for all numbers.

- How many rolls does he tell the truth? 60 rolls ( $ = \frac{2}{3} \times 90 $ )

  - Of those 60, how many rolls are 3? 10 rolls ( $ = \frac{1}{6} \times 60 $ )

    - Of those 10, how many rolls does he call as "3"? These are the number of rolls where the die is 3 and he says "3": **10 rolls** ( $ = \frac{10}{10} \times 10 $ )

- How many rolls does he lie? 30 rolls ( $ = \frac{1}{3} \times 90 $ )

  - Of those 30 rolls, how many are 3? These are the number of rolls where the die is 3, but he will NOT say "3": 5 rolls ( $ = \frac{1}{6} \times 30 $ )

  - Of those 30 rolls, how many are NOT 3? These are the number of rolls where he may or may not say "3": 25 rolls ( $ = \frac{5}{6} \times 30 $ )

    - ***Assuming never a 3***, of those 25 rolls, how many does he call as 3? These are the number of rolls where the die is something other than a three, but he says "3": **0 rolls** ( $ = \frac{0}{5} \times 25 $ )

- Of the 90 rolls, on how many rolls does he say "3"? 10 = 10 truth + 0 lies rolls

- Of those 10 rolls, how many are actually a 3? 10

- 10/10 = 1
