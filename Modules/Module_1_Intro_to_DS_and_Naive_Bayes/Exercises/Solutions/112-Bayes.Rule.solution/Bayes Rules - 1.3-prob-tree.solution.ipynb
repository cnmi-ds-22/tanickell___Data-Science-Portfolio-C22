# Problem 1


Joan says she is itchy. There is a test for Allergy to Cats, but this test is not always right:

* For people that really do have the allergy, the test says "Yes" 80% of the time
* For people that do not have the allergy, the test says "Yes" 10% of the time

If 1% of the population has the allergy, and Joan's test says "Yes", what are the chances that Joan really has the allergy?

## Probability tree with arbitrary start number


Assume a population of 10,000 people.
- 1% have the allergy, i.e. 100 people ( $ = 10,000 \times 1\% $ )
  - Of the 100 people with the allergy, 80 test positive ( $ = 100 \times 80\% $ )
  - Of the 100 people with the allergy, 20 test negative ( $ = 100 - 80 $ )

- 99% do not have the allergy, i.e. 9,900 people ( $ = 10,000 - 100 $ )
  - Of the 9,900 people without the allergy, 990 test positive ( $ = 9,900 \times 10\% $ )
  - Of the 9,900 people without the allergy, 8,910 test negative ( $ = 9,900 - 990 $ )

- How many have positive test? 1,070 ( $ = 80 + 990 $ )
- Of the 1,070, how many have the allergy? 80



#### Calculate directly using counts.

```python
80/1070
```

- $ \approx 7.5\% $
