## Problem 2



Write a Python program that takes a word and reverses it.

Input:

input_word = 'stressed'

Output:

desserts

Solution:

```python
# Solution - variant
import numpy as np

input_word = 'stressed'
print( "".join(np.flip(np.fromiter(input_word, dtype='U1'))) )
```

## Problem 4



Write a Python program to construct the following pattern:



@

@@

@@@

@@@@

@@@@@

@@@@

@@@

@@

@


**Hint:** `@*3` will print out `@@@`.


```python
# Solution
import numpy as np

n = 5

lines = np.char.multiply('@', np.r_[1:n+1, n-1:0:-1])

print('\n'.join(lines))
```

```python
# Solution - variant
import numpy as np

n = 5

x = np.arange(1, n * 2)
result = np.minimum(x, n * 2 - x)
tree = np.char.multiply("@", result)
print("\n".join( tree ) )
```

```python
# Solution - variant
import numpy as np

n = 5
tree = np.char.multiply("@", n - abs(np.arange(1-n,n)))
print("\n".join( tree ) )
```

# Problem 5



Write a Python program which iterates through the integers from 0 to 50. For multiples of three print "Fizz" instead of the number and for the multiples of five print "Buzz". For numbers which are multiples of both three and five print "FizzBuzz".  

For this problem zero is considered a multiple of three and five 3\*0=5\*0=0

Sample Output :  
Fizz Buzz

1

2

Fizz

4

Buzz

Fizz

7

8

Fizz

Buzz

<etc.>

```python
# Solution - variant
import numpy as np

n = 51

memo = np.arange(n).astype("str")
memo[::3] = ["Fizz"] * len(memo[::3])
memo[::5] = ["Buzz"] * len(memo[::5])
memo[::15] = ["FizzBuzz"] * len(memo[::15])
print("\n".join( memo ))
```

# Problem 6



Write a Python program to calculate a dog's age in dog's years.

Note: For the first two calendar years, a dog year is equal to 10.5 calendar years. After that, each dog year equals 4 calendar years.

Input:

calendar_age = 5

Preferred Output:

The dog's age in dog's years is 33

```python
# Solution
# first 2 years. calendar year = 10.5 dog years
# after that calendar year = 4 dog years

calendar_age = 5
dog_age = 0

for i in range(1, calendar_age+1) :
  if i <= 2 :
    dog_age += 10.5
  else :
    dog_age += 4

print(f"The dog's age in dog's years is {dog_age}")
```
