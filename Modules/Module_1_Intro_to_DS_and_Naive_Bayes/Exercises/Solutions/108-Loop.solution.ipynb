## Problem 1

Write a Python program that prints out the numbers from 2 to 60 in increments of two

Output: 2 4 6 ... 56 58 60

```python
# Solution - variant
for i in range(2, 61, 2):
  print(i, end= " ")
```

Solution:

```python
# Solution - variant
input_word = 'stressed'
print(input_word[::-1])
```

Write a Python program that prints each item and its corresponding type from the following list.

Input:

datalist = [1452, 11.23, 1+2j, True, 'w3resource', (0, -1), [5, 12],
{"class":'V', "section":'A'}]

Preferred Output:

Type of  1452  is  <class 'int'>                                                                              
Type of  11.23  is  <class 'float'>                                                                            
Type of  (1+2j)  is  <class 'complex'>                                                                         
Type of  True  is  <class 'bool'>

Type of  (0, -1)  is  <class 'tuple'>                                                                         
Type of  [5, 12]  is  <class 'list'>                                                                          
Type of  {'class': 'V', 'section': 'A'}  is  < class 'dict'>

Solution:

```python
datalist = [1452, 11.23, 1+2j, True, 'w3resource', (0, -1), [5, 12], {"class":'V', "section":'A'}]
datalist
```

```python
# Solution - variant
for data in datalist:
  print(f"Type of {data} is {type(data)}")
```

Write a Python program to construct the following pattern:

@@@@

@@@@@

@@@@

@@@

@@

@


**Hint:** `@*3` will print out `@@@`.


```python
# Solution - variant
for i in [1, 2, 3, 4, 5, 4, 3, 2, 1]:
  print('@'*i)
```

```python
# Solution - variant
for i in list(range(1,6)) + list(range(4,0,-1)):
  print('@'*i)
```

```python
# Solution - variant
limit = 5
for i in range(1, limit * 2):
  print( '@'*min(i, limit * 2 - i) )
```

```python
# Solution = variant
limit = 5
for i in range(1-limit,limit):
  print("@"*(limit-abs(i)))
```

```python
# Solution - variant
limit = 5
print("\n".join( "@"*(limit-abs(i)) for i in range(1-limit,limit)) )
```

# Problem 5



Write a Python program which iterates through the integers from 0 to 50. For multiples of three print "Fizz" instead of the number and for the multiples of five print "Buzz". For numbers which are multiples of both three and five print "FizzBuzz".  

For this problem zero is considered a multiple of three and five 3\*0=5\*0=0

Sample Output :  
Fizz Buzz

1

2

Fizz

4

Buzz

Fizz

7

8

Fizz

Buzz

<etc.>

```python
# Solution - variant 1
for i in range(0,51):
  if i % 3 == 0 and i % 5 == 0:
    print("FizzBuzz")
  elif i % 3 == 0:
    print("Fizz")
  elif i % 5 == 0 :
    print("Buzz")
  else :
    print(i)
```

```python
# Solution - variant
for i in range(0,51):
  output = "Fizz" if not i % 3 else ""
  output += "Buzz" if not i % 5 else ""
  print(f"{output if output else i }")
```

```python
# Solution - variant
memo = list(map(str,range(51)))
# memo = [ str(i) for i in range(51) ]  # same using list comprehension
memo[::3] = ["Fizz"] * len(memo[::3])
memo[::5] = ["Buzz"] * len(memo[::5])
memo[::15] = ["FizzBuzz"] * len(memo[::15])
print("\n".join( memo ))
```

```python
# Solution - variant
memo_dict  = { i:"Fizz"     for i in range(0,15, 3) }
memo_dict |= { i:"Buzz"     for i in range(0,15, 5) }
memo_dict |= { i:"FizzBuzz" for i in range(0,15,15) }
# print(memo_dict)
for i in range(51):
  print(memo_dict.get(i%15,i))

# same using list comprehension
# print("\n".join( memo_dict.get(i%15,str(i)) for i in range(51) ) )
```

# Problem 6



Write a Python program to calculate a dog's age in dog's years.

Note: For the first two calendar years, a dog year is equal to 10.5 calendar years. After that, each dog year equals 4 calendar years.

Input:

calendar_age = 5

Preferred Output:

The dog's age in dog's years is 33

```python
# Solution
# first 2 years. calendar year = 10.5 dog years
# after that calendar year = 4 dog years

calendar_age = 5
dog_age = 0

for i in range(1, calendar_age+1) :
  if i <= 2 :
    dog_age += 10.5
  else :
    dog_age += 4

print(f"The dog's age in dog's years is {dog_age}")
```

```python
# Solution - variant
# first 2 years. calendar year = 10.5 dog years
# after that calendar year = 4 dog years

calendar_age = 5
dog_age = 0

for i in range(1, calendar_age+1):
  dog_age += 10.5 if i <= 2 else 4

print(f"The dog's age in dog's years is {dog_age}")
```
