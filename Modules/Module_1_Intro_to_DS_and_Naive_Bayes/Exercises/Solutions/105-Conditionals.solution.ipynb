## Problem 1

## Problem 5



Make a list called mypets with your pets (real or imaginary) as elements.  
- if mypets contains 'dog' and 'cat', print 'I am a dog person and a cat person'
- else if mypets contains 'dog', print 'I am a dog person'
- else if mypets contains 'cat', print 'I am a cat person'
- otherwise, print 'I am not a cat person or a dog person'

```python
mypets = '''
  dog
  cat
'''.split()
mypets
```

```python
# Using a "in" test
if ( ( "dog" in mypets ) and ( "cat" in mypets ) ):
  print("I am a dog person and a cat person")
elif ( "dog" in mypets ):
  print("I am a dog person")
elif ( "cat" in mypets ):
  print("I am a cat person")
else:
  print("I am not a cat person or a dog person")
```

```python
# Using sets
mypets = set(mypets)
if ( {"dog", "cat"} - mypets == set() ):
  print("I am a dog person and a cat person")
elif ( {"dog"} - mypets == set() ):
  print("I am a dog person")
elif ( {"cat"} - mypets == set() ):
  print("I am a cat person")
else:
  print("I am not a cat person or a dog person")
```

```python
# Using sets and len() instead of equality
mypets_set = set(mypets)
if ( not len({"dog", "cat"} - mypets_set) ):
  print("I am a dog person and a cat person")
elif ( not len({"dog"} - mypets_set) ):
  print("I am a dog person")
elif ( not len({"cat"} - mypets_set) ):
  print("I am a cat person")
else:
  print("I am not a cat person or a dog person")
```

```python
# Using match case
match ( ( "dog" in mypets ), ( "cat" in mypets ) ):
  case ( True, True ):
    print("I am a dog person and a cat person")
  case ( True, False ):
    print("I am a dog person")
  case ( False, True ):
    print("I am a cat person")
  case ( False, False ):
    print("I am not a cat person or a dog person")
  case (_):
    print("huh?")
```

```python
# Using the itertools module to help test all possible combinations
from itertools import chain, combinations

mypets = ['dog', 'cat']
all_combos_of_mypets = chain.from_iterable( combinations(mypets, r) for r in range(len(mypets) + 1) )

for mp in all_combos_of_mypets:
  print(f"{mp=}")
  a = ( "dog" in mp )
  b = ( "cat" in mp )

  if ( a and b ):
    print("I am a dog person and a cat person")
  elif ( a ):
    print("I am a dog person")
  elif ( b ):
    print("I am a cat person")
  elif ( not ( a and b ) ):
    print("I am not a cat person or a dog person")
  else:
    print("huh?")

  print()

```

```python
# Using the itertools module to help test all possible combinations
from itertools import chain, combinations

mypets = ['dog', 'cat']
all_combos_of_mypets = chain.from_iterable( combinations(mypets, r) for r in range(len(mypets) + 1) )

for mp in all_combos_of_mypets:
  print(f"{mp=}")
  a = ( "dog" in mp )
  b = ( "cat" in mp )

  match (a, b):
    case ( True, True ):
      print("I am a dog person and a cat person")
    case ( True, False ):
      print("I am a dog person")
    case ( False, True ):
      print("I am a cat person")
    case ( False, False ):
      print("I am not a cat person or a dog person")
    case (_):
      print("huh?")
  print()

```

```python
# Using a dictionary lookup
# Using the itertools module to help test all possible combinations
mypets = ['dog', 'cat']
all_combos_of_mypets = chain.from_iterable( combinations(mypets, r) for r in range(len(mypets) + 1) )

responses = {
    ( True,  True  ) : "I am a dog person and a cat person",
    ( True,  False ) : "I am a dog person",
    ( False, True  ) : "I am a cat person",
    ( False, False ) : "I am not a cat person or a dog person"
}

for mp in all_combos_of_mypets:
  print(f"{mp=}")
  a = ( "dog" in mp )
  b = ( "cat" in mp )
  print(responses[( a, b )])
  print()
```
