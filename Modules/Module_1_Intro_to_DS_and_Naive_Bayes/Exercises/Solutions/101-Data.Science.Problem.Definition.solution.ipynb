# Data Science Problem Definition


For each item develop a well-worded data science problem statement.  Your answer should include what type of data science problem this is ( supervised vs unsupervised ) and then whether it is a classification, regression, or clustering problem.  If it is a supervised problem, you should state what the target of the prediction is.

---

1. Zillow wants you to predict real estate sales prices. They have a large
database of 12 years of home features accompanied by the corresponding
sales prices.



Characteristics:

1. target: home price
1. supervised
1. regression

Statement:

- Using Zillow's 12-year history of home prices, we will use a supervised learning regression model to predict home prices.



---

2. Amazon wants you to develop a code that upsells (add additional items to the
customer’s order) based on what items have been viewed and bought by that
customer over the past 2 years.



Characteristics:

1. target: none
1. unsupervised
1. clustering

Statement:

- Using Amazon's viewing and order history for a customer from the past 2 years, we will use a clustering algorithm to find associated items to suggest for the customer.

---

3. MGM wants to know what features of a movie seem to increase the movie’s
box office income. They have a large database of previously released movies
accompanied by the resulting box office revenue.



Characteristics:

1. target: box office revenue
1. supervised
1. regression

Statement:

- We will use a supervised regression model to predict box office revenue based on features of previously released movies.

---

4. Spotify wants you to develop a method of predicting whether a customer is
about to terminate their service BEFORE they terminate so that they can offer
enticements to the customer to keep the service and not terminate (called
‘churn’)



Characteristics:

1. target: terminate
1. supervised
1. classification

Statement:

- Using historical data of customers usage patterns and knowing which customers terminated and did not terminate their service, we will build a supervised classification model to predict where or not a customer terminate will terminate.


---

5. Comcast wants you to develop a chat bot (automated customer service) to
take text input and respond with the appropriate help and/or service
response.



Characteristics:

1. target: chat bot response
1. supervised
1. classification

Statement:

- We will build a chatbot to generate responses using a supervised classification model based on historical inputs with known outputs.

---

6. Instagram wants you to analyze images and output whether the image
contains people, animals, scenery, or some other subject.


Characteristics:

1. target: image subject
1. supervised
1. classification

Statement:

- Using archives of labeled images, we will build a supervised classification model to identify people, animals, scenery, and other subjects.
