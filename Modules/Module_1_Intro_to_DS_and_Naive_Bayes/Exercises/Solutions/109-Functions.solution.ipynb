## Problem 1

**Part A:** Write a function that takes a number as input and counts up to that number in increments of 2.  (Output should be done using print().  The function does not need to return anything.)

For example:
Input: 60  

Output: 2 4 6 ... 56 58 60

````python
def count_by_twos(number: int) -> None:
  """Print output that counts by two up to the given number"""
  for i in range(2, number + 1, 2):
    print(i, end= " ")
````

## Problem 2



Write a Python function that takes a word and reverses it.  Function should return the result with a `return` statement


Function Input:

input_word = 'fantastic'

Returns:

citsatnaf

Solution:

````python
# Solution
def reverse_word(word: str) -> str:
  """Takes a word and returns it with the letter in reverse order."""
  return word[-1::-1]
````

````python
input_word = 'fantastic'
reverse_word(input_word)
````

Write a Python function that takes a list as input and prints each item and its corresponding type from the following list.

Input:

datalist = [1452, 11.23, 1+2j, True, 'w3resource', (0, -1), [5, 12],
{"class":'V', "section":'A'}]

Preferred Output:

Solution:

````python
my_list = [1452, 11.23, 1+2j, True, 'w3resource', (0, -1), [5, 12], {"class":'V', "section":'A'}]
````

## Problem 4



Write a Python function to construct the following pattern.  The input will be a positive integer that determines the line with the maximum number of '@' symbols.  

Input: 5  
Output:

@

@@

@@@

@@@@

@@@@@

@@@@

@@@

@@

@


Solution:

````python
# Solution
def make_arrow(max_symbols: int = 5, symbol: str = '@') -> None:
  """ Make an arrow shape of symbols.  Line with max number of symbols has max_symbols."""
  for i in list(range(1,max_symbols + 1)) + list(range(max_symbols - 1,0,-1)):
    print(i*symbol)
````

````python
make_arrow(5)
````

````python
make_arrow(7, '&')
````

# Problem 5



Write a Python function that takes a positive integer (n) as input and iterates through the integers from 0 to n.

Return a list that has
-  the number if it is not a multiple of 3 or 5
-  'Fizz' if the number is a multiple of 3
-  'Buzz' if the number is a multiple of 5
-  'FizzBuzz' if the number is a multiple of 3 and 5

```python
Sample Input: 7

Sample Return:
['FizzBuzz', '1', '2', 'Fizz', '4', 'Buzz', 'Fizz', '7']
```


````python
# Solution
def fizzbuzz(number: int) -> list:
  """ Make a fizzbuzz list"""

  memo = [ str(i) for i in range(number+1) ]
  memo[::3] = ["Fizz"] * len(memo[::3])
  memo[::5] = ["Buzz"] * len(memo[::5])
  memo[::15] = ["FizzBuzz"] * len(memo[::15])
  return memo
````

````python
fizzbuzz(30)
````

# Problem 6



Write a Python function to calculate a dog's age in dog's years.  (Input is human age).

Note: For the first two years, a dog year is equal to 10.5 human years. After that, each dog year equals 4 human years.

Input:

h_age = 5

Returns the string:

The dog's age in human's years is 33

````python
# Solution
def calculate_dogs_years(human_age: float) -> str :
  """ Calculate dog years from human years"""
  if human_age <= 2 :
    dog_years = max(human_age,0) * 10.5
    return f"The dog's age in human's years is {dog_years:.1f}."
  else:
    dog_years = 21 + 4 * (human_age-2)
    return f"The dog's age in human's years is {int(dog_years)}."
````

````python
for i in list(range(-1,6)) + list(range(6,41,2)):
  years = i/2
  print(f"{years:5}, {calculate_dogs_years(years)} ")
````
