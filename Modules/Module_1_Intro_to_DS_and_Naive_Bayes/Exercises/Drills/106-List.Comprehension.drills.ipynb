## Part 1: Numbers (Integers & Floats)

```python
# 1. Doubling integers
numbers = [1, 2, 3, 4, 5]
doubled = [x * 2 for x in numbers]
doubled
```

```python
doubled = []
for i in numbers:
  doubled.append(i*2)

doubled
```

```python
# Your for-loop here:
doubled = []
for x in numbers:
  doubled.append( x * 2 )
doubled
```

```python
# 2. Squaring numbers
data = [1, 3, 5, 7, 9]
squares = [num ** 2 for num in data]
squares
```

```python
squares = []
for datum in data:
  squares.append(datum ** 2)
squares
```

```python
# 3. Converting Celsius to Fahrenheit
celsius = [0.0, 10.5, 20.0, 35.6]
fahrenheit = [(temp * 9/5) + 32 for temp in celsius]
fahrenheit
```

```python
fahrenheit = []
for temp in celsius:
  fahrenheit.append((temp * 9/5) + 32)
fahrenheit
```

```python
# 4. Filtering for even numbers
scores = [12, 15, 18, 21, 24, 27, 30]
evens = [n for n in scores if n % 2 == 0]
evens
```

```python
evens = []
for score in scores:
  if (score % 2 == 0):
    evens.append(score)
evens
```

```python
# 5. Keeping only positive balances
balances = [100.50, -20.00, 50.25, -5.10, 0.0]
positive_only = [bal for bal in balances if bal > 0]
positive_only
```

```python
positive_only = []
for balance in balances:
  if balance > 0:
    positive_only.append(balance)
positive_only
```
