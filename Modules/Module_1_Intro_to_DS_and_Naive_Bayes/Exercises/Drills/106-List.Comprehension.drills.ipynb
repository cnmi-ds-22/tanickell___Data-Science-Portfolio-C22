## Part 1: Numbers (Integers & Floats)

```python
# 1. Doubling integers
numbers = [1, 2, 3, 4, 5]
doubled = [x * 2 for x in numbers]
doubled
```

```python
# Your for-loop here:
numbers = [1, 2, 3, 4, 5]
doubled = []
for x in numbers:
  doubled.append( x * 2 )
doubled
```

```python
# 2. Squaring numbers
data = [1, 3, 5, 7, 9]
squares = [num ** 2 for num in data]
squares
```

```python
# Your for-loop here:
data = [1, 3, 5, 7, 9]
squares = []
for num in data:
  squares.append(num ** 2 )
squares
```

```python
# 3. Converting Celsius to Fahrenheit
celsius = [0.0, 10.5, 20.0, 35.6]
fahrenheit = [(temp * 9/5) + 32 for temp in celsius]
celsius
```

```python
# Your for-loop here:
celsius = [0.0, 10.5, 20.0, 35.6]
fahrenheit = []
for temp in celsius:
  fahrenheit.append((temp * 9/5) + 32)
celsius
```

```python
# 4. Filtering for even numbers
scores = [12, 15, 18, 21, 24, 27, 30]
evens = [n for n in scores if n % 2 == 0]
evens
```

```python
# Your for-loop here:
scores = [12, 15, 18, 21, 24, 27, 30]
evens = []
for n in scores:
  if n % 2 == 0:
    evens.append(n)
evens
```

```python
# 5. Keeping only positive balances
balances = [100.50, -20.00, 50.25, -5.10, 0.0]
positive_only = [bal for bal in balances if bal > 0]
positive_only
```

```python
# Your for-loop here:
balances = [100.50, -20.00, 50.25, -5.10, 0.0]
positive_only = []
for bal in balances:
  if bal > 0:
    positive_only.append(bal)
positive_only
```

## Part 2: Strings

```python
# 6. Uppercasing names
names = ["alice", "bob", "charlie"]
shouting = [name.upper() for name in names]
shouting
```

```python
# Your for-loop here:
names = ["alice", "bob", "charlie"]
shouting = []
for name in names:
  shouting.append(name.upper())
shouting
```

```python
# 7. Getting string lengths
words = ["apple", "banana", "cherry", "date"]
lengths = [len(w) for w in words]
lengths
```

```python
# Your for-loop here:
words = ["apple", "banana", "cherry", "date"]
lengths = []
for w in words:
  lengths.append(len(w))
lengths
```

```python
# 8. Filtering by initial letter
animals = ["cat", "dog", "cheetah", "elephant", "camel"]
c_animals = [animal for animal in animals if animal.startswith("c")]
c_animals
```

```python
# Your for-loop here:
animals = ["cat", "dog", "cheetah", "elephant", "camel"]
c_animals = []
for animal in animals:
  if animal.startswith("c"):
    c_animals.append(animal)
c_animals
```

## Part 3: Mixed & Combined Concepts

```python
# 9. Finding long words
fruits = ["kiwi", "mango", "watermelon", "fig", "pineapple"]
long_fruits = [f for f in fruits if len(f) > 5]
long_fruits
```

```python
# Your for-loop here:
fruits = ["kiwi", "mango", "watermelon", "fig", "pineapple"]
long_fruits = []
for f in fruits:
  if len(f) > 5:
    long_fruits.append(f)
long_fruits
```

```python
# 10. Halving numbers if they are large
weights = [10.0, 50.5, 4.2, 100.0, 12.8]
halved_large = [w / 2 for w in weights if w > 15.0]
halved_large
```
