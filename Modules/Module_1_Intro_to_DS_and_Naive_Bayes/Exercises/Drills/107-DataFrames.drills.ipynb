# Dataframe Drills: Characterizing a Data Frame

Different machine learning ( ML ) algorithms have many overlapping requirements and also several different requirements of the data.  For example, all ML algorithms require that none of the features have null values.  And many ML algorithms require that continuous variables be normally distributed.  Therefore, it is necessary to identify which features have nulls ( and address the nulls ) and which feature are continuous.  This exercise provides practice in characterizing a data frame.

## Setup


```python
# @title Create the dictionary of data frames.
import pandas as pd
import numpy as np

# Read in main data frame
url = "https://ddc-datascience.s3.amazonaws.com/Projects/Project.2-Housing/Data/Housing.Data.csv"
df_all = pd.read_csv( url )

# Create a dictionary of data frames
dict_df = {}
np.random.seed(42)
i = 0
dict_df[f"df_{i:03}"] = df_all.sample( n=10 ).sample( n=10, axis = "columns" )
for i in range(1,51):
  row_pick = max( min(200, 0.1 * df_all.shape[0]), 100 )
  nrows = np.random.choice( range( 10, row_pick ) )
  ncols = np.random.choice( range( max( 10, df_all.shape[1] ) ) )
  dict_df[f"df_{i:03}"] = df_all.sample( n=nrows ).sample( n=ncols, axis = "columns" )

# for k, df in df_dict.items():
#   print( k, df.shape )
```
