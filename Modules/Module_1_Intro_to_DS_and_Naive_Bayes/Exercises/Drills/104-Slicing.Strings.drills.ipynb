## Your turn

```python
def solve(i):

  backwards = False

  start = line.find(outputs[i])
  if start == -1:
     backwards = True
     find_index = line[::-1].find(outputs[i])
     start = len(line) - 1 - find_index

  step = -1 if backwards else 1

  end = start + step * len(outputs[i])
  if end == -1:
    end = (len(line) + 1) * -1

  x = slice(start, end, step)

  print(x)
  print(f"outputs[{i}]:")
  print(outputs[i])
  print(line[x])
  print()
```

```python
for i in range(0, len(outputs)):

  solve(i)
```
