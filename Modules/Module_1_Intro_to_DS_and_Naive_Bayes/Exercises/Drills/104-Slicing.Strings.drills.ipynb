```python
for i in range(0, 20):
  start = line.find(outputs[i])
  if start == -1:
    start = len(line) - 1 - line[::-1].find(outputs[i])
    step = -1
    stop = start - len(outputs[i])
    if stop == -1:
      stop = None
  else:
    step = 1
    stop = start + len(outputs[i])

  x = slice(start,stop,step)
  print(i)
  print(x)
  print(outputs[i])
  print(line[x])
  print()
```
