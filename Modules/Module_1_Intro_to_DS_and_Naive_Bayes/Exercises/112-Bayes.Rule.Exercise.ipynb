## Problem 1:

Joan says she is itchy. There is a test for Allergy to Cats, but this test is not always
right:

- For people that really do have the allergy, the test says "Yes" 80% of the time
- For people that do not have the allergy, the test says "Yes" 10% of the time

If 1% of the population has the allergy, and Joan’s test says "Yes", what are the chances that Joan really has the allergy?


```python
p_a = 0.01
p_a
```

```python
p_b = 0.8 * p_a + 0.1 * (1 - p_a)
p_b
```

```python
p_ba = 0.8
p_ba
```

```python
p_ab = p_ba * p_a / p_b
p_ab
```

```python
# Joan has about a 7.5% chance of actually being allergic given a positive test.
```

## Problem 3:

An artist competition has entries from three painters: Pam, Paul and Peter.

- Pam put in 15 paintings, 4% of her works have won First Prize.
- Paul put in 5 paintings, 5% of his works have won First Prize.
- Peter put in 10 paintings, 3% of his works have won First Prize.

What is the probability that Pam will win First Prize?

```python
# A = Painting [ Pam, Peter, Paul ]
# B = Win      [ yes, no ]

# P(B = yes, A = Pam)   = 0.04
# P(B = yes, A = Paul)  = 0.05
# P(B = yes, A = Peter) = 0.03
```

```python
p_ba = 0.04
p_a  = 0.5
p_b = ((0.5 * 0.04) + (0.166 * 0.05) + (0.333 * 0.03))

p_ba, p_a, p_b
```

```python
p_ab = p_ba * p_a / p_b
p_ab
```

```python
# There is a 52.2% chance that given a winning painting, it belongs to Pam.
```

## Problem 4: (Challenge)

A man speaks the truth 2 out of 3 times. He throws one die and says that number
obtained is a three. Find the probability that the number obtained is actually a three.

```python
# A = Truth          --> [ yes, no ]
# B = Roll(reported) --> [ 1, 2, 3, 4, 5, 6 ]

# Given a three is reported, find the probability that the man is truthful.
# Alt: find the probability of truth given 3.

# Assume a total of 360. Then:

p_ba = 40 / 240
p_a  = 2 / 3
p_b  = ((1/6) * (2/3)) + ((1/6) * (1/3))

p_ba, p_a, p_b
```

```python
p_ab = p_ba * p_a / p_b
p_ab
```

```python
# There is a 66% chance that given a 3 was reported, the man is being truthful,
# and that a 3 was actually obtained.
```
