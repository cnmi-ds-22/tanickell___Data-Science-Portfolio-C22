## Problem 2



Write a Python program that takes a word and reverses it.

Input:

```
input_word = 'stressed'
```

Output:
```
desserts
```

Solution:

````python
word = "stressed"
my_array  = np.array(list(word))
rev_array = my_array[::-1].copy()

"".join(list(rev_array))
````

````python
def reverse(word):
  '''
  Takes a word (string) and reverses it. Uses numpy arrays.
  '''
  my_array  = np.array(list(word))
  rev_array = my_array[::-1].copy()

  return "".join(list(rev_array))
````

````python
word = "stressed"
word_reverse = reverse(word)

word_reverse
````

# Problem 5



Write a Python program which iterates through the integers from 0 to 50. For multiples of three print "Fizz" instead of the number and for the multiples of five print "Buzz". For numbers which are multiples of both three and five print "FizzBuzz".  

For this problem zero is considered a multiple of three and five 3\*0=5\*0=0

Sample Output :  
```
Fizz Buzz
1
2
Fizz
4
Buzz
Fizz
7
8
Fizz
Buzz
<etc.>
```


````python
f   = 3
b   = 5
fb  = 15
num = 50

for i in range (0, num + 1):
  if i % f == 0 and i % b == 0:
    print("Fizz Buzz")
  elif i % f == 0:
    print("Fizz")
  elif i % b == 0:
    print("Buzz")
  else:
    print(i)
````

# Problem 6



Write a Python program to calculate a dog's age in dog's years.

Note: For the first two calendar years, a dog year is equal to 10.5 calendar years. After that, each dog year equals 4 calendar years.

Input:
```
calendar_age = 5
```

Preferred Output:

```
The dog's age in dog's years is 33
```

````python
calendar_age = 5

````

````python
calendar_age = 5

if calendar_age <= 2:
  dog_age = 2 * 10.5
else:
  dog_age = 2 * 10.5 + (calendar_age - 2) * 4

print(f"The dog's age in dog's years is {int(dog_age)}")
````
