## Rock, Paper, Scissors

```python
rng = np.random.default_rng()
```

```python
rng = np.random.default_rng()
game_over = False
game_count = 0

wins   = 0
losses = 0
draws  = 0

moves = np.array(["rock", "paper", "scissors"])

print(f"Welcome to Python Rock Paper Scissors!")

# Game Loop
while not game_over:

  # increment game count
  game_count += 1
  user_choice = ""

  # validate user input
  while user_choice not in ["1", "2", "3"]:
    # get user input
    user_choice = input(
      "Please choose (1) rock, (2) paper, or (3) scissors: --> "
    )
    # if user_choice not in [1, 2, 3]:
    #   print("Choice must be 1, 2, or 3. Try again.")

  # convert user choice to int
  user_choice = int(user_choice)

  # normalize to zero-indexing
  user_choice -= 1

  #generate computer move
  comp_choice = rng.integers(0, 3)

  result = ""
  if   comp_choice == (user_choice - 1) % 3:
    result = "win"
    wins += 1
  elif comp_choice == user_choice:
    result = "draw"
    draws += 1
  elif comp_choice == (user_choice + 1) % 3:
    result = "lose"
    losses += 1

  print(f"You chose:      {moves[user_choice]}")
  print(f"Computer chose: {moves[comp_choice]}")
  print(f"Result: {result}")
  print(f"wins: {wins}, losses: {losses}, draws: {draws}")
  print()

  # play again?
  play_again_str = ""
  while play_again_str not in ["y", "n"]:
    play_again_str = input("Play again? y/n: --> ")
    play_again_str = play_again_str.lower()
    play_again = play_again_str == "y"

  if not play_again:
    game_over = True

print("Thanks for playing!")
```

## Number Guesser

```python
rng = np.random.default_rng()
attempts = 0
winner = False
play_again = True

target = rng.integers(1, 11)

print(f"Welcome to Python Number Guesser!")

while play_again:

  winner = False
  attempts = 0
  print("I'm thinking of a number between 1 and 10.")

  while attempts < 5 and not winner:
    user_guess = input("Please guess my number: --> ")
    if user_guess not in [str(x) for x in list(range(1, 11))]:
      print("Please guess a number between 1 and 10.")
      continue

    user_guess = int(user_guess)
    attempts += 1

    if user_guess > target:
      print("Too high. Guess again!")
    elif user_guess < target:
      print("Too low. Guess again!")
    else:
      print(f"BINGO! You win! You got it in {attempts} guesses :D")
      winner = True

  if not winner:
    print("Too bad, you're out of guesses. You lose :/")

  # play again?
  response = ""
  while response not in ["y", "n"]:
    response = input("Play again? y/n: --> ")
    response = response.lower()
    play_again = response == "y"

  if not play_again:
    game_over = True

print("Thanks for playing!")
```
