##  Basic

1. Given	a list,	remove the	element	at	index	number 4 and add	it at index position 2 (over-writing what is currently there) and also add it at the end of the list.

```python
list1 = [34, 56, 49, 4900, 249, 494]
```

````
# Declare numpy array version of list1
array1 = np.array([34, 56, 49, 4900, 249, 494])
array1
````

````
# Save a copy of element at index 4
remove_element = array1[4]
remove_element, array1[4]
````

````
# Check on array
array1
````

````
# Remove the element at index 4
array1 = np.delete(array1, 4)
array1
````

````
# Overwrite position 2 element with removed element
array1[2] = remove_element
array1
````

````
# Finally, make new array with removed element appended to end
array1 = np.append(array1, remove_element)
array1
````

2. Create a dictionary to store some information about yourself.  For example you might have key value pairs like:
- favorite ice cream & chocolate
- hair color & blond

````
# Represent dictionary as numpy array of two numpy arrays
me_keys = np.array(['name', 'hair color', 'favorite dessert'])
me_vals = np.array(['tim', 'brown', 'tiramisu'])
me_dict = np.array([me_keys, me_vals])

me_dict
````

3. Add an additional piece of information about yourself to your dictionary from above.

````
me_dict = np.array([
    np.append(me_dict[0], 'favorite movie'),
    np.append(me_dict[1], 'Phantom Thread')
])

me_dict
````

4. Remove one of the entries from your dictionary

````
remove_index = np.where(me_dict[0] == 'hair color')
me_dict = np.array([
    np.delete(me_dict[0], remove_index),
    np.delete(me_dict[1], remove_index)
])

me_dict
````

5. Create your own tuple.  You can put whatever you want in it.  

````
# tuple heterogeneity of data type and immutability
numpy_tuple_structure = np.dtype([
    ('name', 'U10'),
    ('num_cats', 'i4'),
    ('favorite_dessert', 'f8')
])
````

````
type(numpy_tuple_structure)
````

````
numpy_tuple = np.array(
    ('tim', 3, 3.14), # no [] around the parenthesized group
    dtype=numpy_tuple_structure
)
````

````
numpy_tuple
````

````
# numpy_tuple + ('bob', 2, 3.0)
````
