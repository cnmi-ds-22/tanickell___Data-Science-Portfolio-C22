## Excel

In Excel, a file consists of a single workbook.  Within a workbook there can be one or more sheets ( sometimes called tabs. )

For example,
- Workbook ( Excel file )
  - Sheet1
  - Sheet2
  - Sheet3

### Read Excel from a URL


```python
# Location of Excel file
dat_file_url = 'http://ddc-datascience.s3.amazonaws.com/animals.xlsx'
dat_file_url
```

```python
# Reading in one sheet from a workbook
animal_dat = pd.read_excel( dat_file_url, sheet_name = "Sheet1")
animal_dat
```

```python
# Reading in multiple sheets from a workbook
workbook = pd.ExcelFile( dat_file_url )

# listing the worksheets
workbook.sheet_names
```

```python
# Reading in a few sheets from a workbook
animals_dat1 = pd.read_excel( workbook, "Sheet1")
animals_dat2 = pd.read_excel( workbook, "Sheet2")
display(animals_dat1)
print()
display(animals_dat2)
```

```python
# Can also read in multiple sheets from a workbook using a loop
with pd.ExcelFile(dat_file_url) as workbook:
  data = {
    sheet_name: pd.read_excel( workbook, sheet_name)
      for sheet_name in workbook.sheet_names
  }

data.keys()
```

```python
data["Sheet1"]
```

```python
data['Sheet2']
```

```python
data['Sheet3']
```

## HTML
Reading in html can be useful if there is a table of data on a website that you want to parse.

### Read HTML from a URL


```python
url = "https://en.wikipedia.org/wiki/New_Mexico"
url = "https://en.wikipedia.org/w/index.php?title=New_Mexico&oldid=1250070720"
url
```

```python
#  Configure a user-agent string
headers = {
  'User-Agent': "pandas"
}
headers
```

```python
# Read all the tables
nm_tables = pd.read_html( url, storage_options=headers )
# nm_tables = pd.read_html( url )
len( nm_tables ) # nm_table contains all of the tables on this webpage
```

```python
nm_tables[5]
```

```python
my_df = nm_tables[5].iloc[0:10,1:4].copy()
my_df
```

```python
my_df.columns[0][0]
```

```python
for i, table in enumerate(nm_tables):
  print(f"Table index: {i}")
  print(table.columns)
  print()
```

```python
# To only get one table, use the match argument
nm_lang_tables = pd.read_html(url, match = 'English only', storage_options=headers)
len(nm_lang_tables)
```

```python
nm_lang_tables
```

```python
nm_lang_df = nm_lang_tables[0]
nm_lang_df.shape
```

```python
nm_lang_df
```

```python
nm_lang_df.info()
```

## Parquet


### Download a large CSV


```python
url = 'https://huggingface.co/datasets/rwcitek/medicare/resolve/main/DAC_NationalDownloadableFile.csv'
url
```

```python
!curl -L -O {url}
```

```python
csv_file = url.split("/")[-1]
csv_file
```

```python
!ls -l {csv_file}
```

### Read from CSV


```python
medicare_df = pd.read_csv( csv_file, dtype="str" )
medicare_df.shape
```

```python
medicare_df.info()
```

```python
medicare_df['Grd_yr'] = medicare_df['Grd_yr'].astype('Int64')
medicare_df.info()
```

### Save to parquet


Notice that parquet is significantly smaller than the CSV.


```python
parquet_file = csv_file.rsplit(".")[0] + ".parquet"
parquet_file
```

```python
medicare_df.to_parquet(parquet_file, index=False)
```

```python
!ls -l --si {csv_file} {parquet_file}
```

### Read from parquet


Notice that parquet preserves the data types of the columns.


```python
mc_df2 = pd.read_parquet( parquet_file )
mc_df2.info()
```
